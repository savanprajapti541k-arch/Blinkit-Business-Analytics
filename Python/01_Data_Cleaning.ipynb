### Data Cleanning

```python
orders.drop_duplicates(inplace=True)
products.drop_duplicates(inplace=True) 
customers.drop_duplicates(inplace=True) 
delivery.drop_duplicates(inplace=True) 
inventory.drop_duplicates(inplace=True) 
feedback.drop_duplicates(inplace=True)
```

```python
orders.columns = orders.columns.str.strip()
products.columns = products.columns.str.strip()
customers.columns = customers.columns.str.strip()
delivery.columns = delivery.columns.str.strip()
inventory.columns = inventory.columns.str.strip()
feedback.columns = feedback.columns.str.strip()
```

```python
orders = orders.apply(lambda col: col.str.strip() if col.dtype == "object" else col)
products = products.apply(lambda col: col.str.strip() if col.dtype == "object" else col)
customers = customers.apply(lambda col: col.str.strip() if col.dtype == "object" else col)
delivery = delivery.apply(lambda col: col.str.strip() if col.dtype == "object" else col)
inventory = inventory.apply(lambda col: col.str.strip() if col.dtype == "object" else col)
feedback = feedback.apply(lambda col: col.str.strip() if col.dtype == "object" else col)
```

```python
orders["order_date"] = pd.to_datetime(orders["order_date"])
customers["registration_date"] = pd.to_datetime(customers["registration_date"])
```

```python
customers["address"] = customers["address"].fillna("Unknown")
```
